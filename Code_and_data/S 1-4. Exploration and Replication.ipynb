### Check if the Romer and Romer shocks have autocorrelation 
Are they actually orthogonal shocks?

```python
# load data and packages
import pandas as pd
from statsmodels.stats.diagnostic import acorr_ljungbox
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
import matplotlib.pyplot as plt

romer = pd.read_excel("RomerandRomerDataAppendix.xls", sheet_name="DATA BY MONTH")
```

```python
#run test
lb_results = acorr_ljungbox(romer["RESID"], lags=[12, 24], return_df=True)
print(lb_results)
```

Yooooo, this is weird. There seems to be significant autocorrelation when we look at 12 and 24 lags. This means that the shocks aren't orthogonal??

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(8, 4))

fig.suptitle("Residual Diagnostics for Romer & Romer Shock Series", fontsize=18, y=1.05)

plot_acf(romer["RESID"], lags=24, zero=False, ax=ax1)
ax1.set_title("Autocorrelation (ACF)")
ax1.set_xlabel("Lags")  # Add x-axis label to ax1

plot_pacf(romer["RESID"], lags=24, zero=False, ax=ax2)
ax2.set_title("Partial Autocorrelation (PACF)")
ax2.set_xlabel("Lags") 

plt.tight_layout()
plt.show()
```
